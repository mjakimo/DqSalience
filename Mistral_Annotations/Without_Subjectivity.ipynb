# Annotations without subjectivity

The annotations systems is based a lot on the subjectivity of each annotator.
Since we will use the annotations to train a model to annotate more, we'll need to avoid this subjectivity.
We choose to put together the following score (labels) for the annotations:
- 0 and 1
- 6, 7 and 8.

```python
import os 
import pandas as pd
```

## Files Paths

```python
files_paths = os.listdir("/Users/axel/Documents/Universite/Cours/Master_Traitement_Automatique_Des_langues/Master_1/Semester_8/Supervised_Project/Corpus/Sub_Corpus_For_Project/Mistral_Annotations")
files_paths = [file for file in files_paths if ("Without_Subjectivity" not in file and ".DS_Store" not in file)]
files_paths.sort()
len(files_paths)
```

```python
def without_subjectivity(files_paths):
    if "HARCHAL" in files_paths:
        df = pd.read_csv(files_paths, sep=";")
        df[df == 1] = 0
        df[df == 6] = 8
        df[df == 7] = 8
        files_paths.replace(".csv", "")
        df.to_csv(f"Without_Subjectivity/{files_paths}_without_subjectivity.csv")
    else:
        df = pd.read_csv(files_paths)
        df[df == 1] = 0
        df[df == 6] = 8
        df[df == 7] = 8
        files_paths.replace(".csv", "")
        df.to_csv(f"Without_Subjectivity/{files_paths}_without_subjectivity.csv")
    return True
```

```python
i = 1
for file in files_paths:
    without_subjectivity(file)
    print(f"{i}: {file} done")
    i+=1
```
