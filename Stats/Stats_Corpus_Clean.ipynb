## Functions to plot our data

```python
def bar_plot_function_Fixed_axis(x, y, xlabel, ylabel, title):
    fig, ax = plt.subplots()
    ax.bar(x, y, width=1, edgecolor="white", linewidth=0.7)
    ax.set(ylim=(0, 650), xlabel=xlabel, ylabel=ylabel, title=title)
    ax.tick_params(axis='x', rotation=90)
    plt.grid()
    plt.show
```

```python
def bar_plot_function(x, y, xlabel, ylabel, title):
    fig, ax = plt.subplots()
    ax.bar(x, y, width=1, edgecolor="white", linewidth=0.7)
    ax.set(ylim=(0, max(y) + (max(y) * 0.1)), xlabel=xlabel, ylabel=ylabel, title=title)
    ax.tick_params(axis='x', rotation=90, labelsize=5)
    plt.grid()
    plt.show
```

# Graphs about the length of the utterances per speakers of all the corpus

This next function read the files that were cleaned. So be carefull about the file path
To get those files, we use the parser that we created to keep only the text from the xml files.

```python
def check_utterances(file):
    nlp = spacy.load("en_core_web_sm")
    dict_speakers_nb_tokens = {}
    with open(file) as f:
        content_file = f.readlines()
        
        while "[" not in content_file[0]:
            content_file.pop(0)
        for line in content_file:
            list_from_line = line.split()
            list_from_line.pop(0)
            speakers = list_from_line.pop(0)
            Ligne_from_file = nlp(line)
            
            if speakers in dict_speakers_nb_tokens:
                dict_speakers_nb_tokens[speakers] = dict_speakers_nb_tokens[speakers] + len(Ligne_from_file)-5
            else:
                dict_speakers_nb_tokens[speakers] = len(Ligne_from_file)-5 #Because 5 elements are in addition that we don't want to take into account
    return (dict_speakers_nb_tokens)

def check_utterances_plot(files_list):
    #to get the filepath of every file. This notebook need to be in the same folder as the files. 
    txt_name_full = []
    for file in files_list:
        txt_name_full.append(file.replace(".xml", "_clean.txt"))
    txt_name_full.sort()
    i = 0
    for fichier in txt_name_full:
        i+=1
        file_path = "Corpus_Clean/"+fichier #The file path with the corpus cleaned txt file.
        print("\n")
        print(file_path)
        print(f"{i}/{len(files_list)}")
        dict_length_utterance = check_utterances(file_path)
        print(dict_length_utterance)
        bar_plot_function(dict_length_utterance.keys(), dict_length_utterance.values(), "Speakers", "Length of utterances", f"Length of utterances for each speaker in {fichier}")
    return(f"The graph about the utterances length for the {files_list} are done") 
```
